```python
from scipy.stats import norm

fare_Y=500
fare_M=200
fare_Q=100

loc_Y=30
scale_Y=5

loc_M=50
scale_M=8

prob_Y=1-(fare_M/fare_Y)
seats_protected_Y=norm.ppf(prob_Y, loc=loc_Y, scale=scale_Y)

combined_loc=loc_Y + loc_M
combined_scale=(scale_Y**2 + scale_M**2)**0.5

prob_combined=1-(fare_Q/fare_M)

total_seats_protected=norm.ppf(prob_combined,loc=combined_loc,scale=combined_scale)

print("---Advanced 3-Class EMSRb OPtimization---")
print(f"1. Seats to protect strictly for Business (Y): {round(seats_protected_Y)}")
print(f"2. Total Seats to lock away for (Business + Standard) before selling promo (Q): {round(total_seats_protected)}")
```
