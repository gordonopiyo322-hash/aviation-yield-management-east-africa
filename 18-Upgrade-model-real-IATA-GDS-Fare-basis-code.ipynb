```python
AIRCRAFT_SEATS = 100
OVERBOOKING_PAD = 5

TOTAL_AUTHORIZATION_LIMIT = AIRCRAFT_SEATS + OVERBOOKING_PAD

tickets_issued = 0
gross_revenue = 0
gds_inventory= {"YOWRAE": 0, "MOWRAE": 0, "QOWRAE": 0}


passenger_queue =[]

for i in range(110):
    passenger_queue.append(("YOWRAE",3))

print(f"---GDS Inventory Authorization Check (max cap:{TOTAL_AUTHORIZATION_LIMIT}---")

for fare_basis,dtd in passenger_queue:
            if tickets_issued >= TOTAL_AUTHORIZATION_LIMIT:
                continue

            if fare_basis == "YOWRAE":
                ticket_fare = 500

            gds_inventory[fare_basis] +=1
            tickets_issued += 1
            gross_revenue += ticket_fare

print(f"--- IATA - Post Flight Commercial Audit Summary---")
print(f"Total Electronic tickets Issued (ETKT): {tickets_issued}/{TOTAL_AUTHORIZATION_LIMIT}")
print(f"Final Flight Revenue Settled: ${gross_revenue}")

extra_cash = OVERBOOKING_PAD * 500
print(f"Overbooking Stra: +${extra_cash} extra revenue generated!")

                  
                  
```
