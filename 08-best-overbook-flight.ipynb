```python
flights=[
    {"flight": "FZ1289", "load": 0.95, "demand": 220},
    {"flight": "FZ1290", "load": 0.40, "demand":80},
    {"flight": "FZ1281", "load": 0.70, "demand":180},
    {"flight": "FZ1282", "load": 0.85, "demand":210},
 ]


class Flight:
    def __init__(self,seats,distance=4500):
        self.seats=seats
        self.distance=distance
        self.pax_count=int(seats*200)

    def fuel_saving_usd(self):
        price_per_kg=0.80
        empty_fuel=self.distance*7.5
        pax_extra_fuel=self.pax_count*100*0.005*(self.distance/100)
        actual_fuel=empty_fuel+pax_extra_fuel
        max_fuel=(self.distance*7.5) +(200*100*0.005)*(self.distance/100)
        saved_by_empty_seats=max_fuel-actual_fuel
        lido_optimization=actual_fuel*0.03
        return int((saved_by_empty_seats + lido_optimization)*price_per_kg)


print("---RM DECISION: BEST FLIGHT TO OVERBOOK---")        
for f in flights:
     plane=Flight(f["load"])
     saving=plane.fuel_saving_usd()

    


     if f["demand"]>200 and saving>1000:
        decision="OVERBOOKING - Demand kubwa & bado tuna-save!"
     elif f["load"]<0.5:
        decision="PROMO -Load ndogo, tuuze bei cheap"
     else:
        decision="HOLD - Usiguse"



    
     print(f"{f['flight']} | Demand {f['demand']} | Saving ${saving} -> {decision}")        
      



    
```
