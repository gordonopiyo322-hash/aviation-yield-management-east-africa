```python
def calc_saving(load,distance=4500):
    empty_fuel=distance*7.5
    pax=int(load*200)
    pax_fuel=pax*100*0.005*(distance/100)
    actual_fuel=empty_fuel + pax_fuel
    max_fuel=(distance*7.5) + (200*100*0.005*(distance/100))
    return int(((max_fuel-actual_fuel) + (actual_fuel*0.03)) *0.08)

def rm_decision(demand,saving,load):
    if demand>150 and saving>100:
        return "OVERBOOK"
    elif load<0.5:
        return "PROMO"
    else:
        return "HOLD"

flights=[
     {"flight": "FZ1289", "load": 0.95, "demand": 220},
     {"flight": "FZ1290", "load": 0.95, "demand": 200},
     {"flight": "FZ1281", "load": 0.50, "demand": 100},
     {"flight": "FZ1283", "load": 0.00, "demand": 80},
     {"flight": "FZ1285", "load": 0.75, "demand": 180}

 ]

for f in flights:
     saving=calc_saving(f["load"])
     decision=rm_decision(f["demand"],saving ,f["load"])
     print(f"{f["flight"]} -> Saving ${saving} -> {decision}")
```
