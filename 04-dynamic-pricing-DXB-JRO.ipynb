```python
base_price=800
load_factor=0.85
season="Highh - Dec 20"

demand_factor=1+(load_factor*0.8)
season_factor=1.2 if "Dec" in season else 1.0
days_to_depature=5
urgency_factor=1.25 if days_to_depature<7 else 1.0

final_price =base_price+ (base_price*0.5*load_factor)+(base_price*0.2)

print(f"Route:DXB-JRO")
print(f"Load:{load_factor*100}%")
print(f"Season:{season} - Factor {season_factor}")
print(f"Urgency:{days_to_depature} days - Factor {urgency_factor}")
print(f"FINAL PRICE: ${final_price:.2f}")

import matplotlib.pyplot as plt
prices= [800,950,1100,final_price,1500]
labels=["Low","Mid","High","Today Dec20","Peak"]
plt.bar(labels,prices,color='gold')
plt.title(f"DXB-JRO Dynamic price - Today ${final_price:.0f}")
plt.ylabel("Price $")
plt.show()
```
