```python
import pandas as pd
import numpy as np

data={
    'flight':['FZ1289','FZ1283','FZ1285'],
    'demand':[130,60,110],
    'capacity':[110,110,110],
    'no_show_rate':[0.1,0.1,0.1],
    'fuel_max':[5000,5000,5000],
    'fuel_actual':[3000,4000,3200]
}

df=pd.DataFrame(data)

df['overbook_limit']=df['capacity']/(1-df['no_show_rate'])
df['extra_seats']=df['overbook_limit']-df['capacity']
df['fuel_saved']=df['fuel_max']-df['fuel_actual']
df['decision']=np.where(df['demand']>df['capacity'],'OVERBOOK', 'HOLD')
print(df)
                                      

    
    
```

```python
import matplotlib.pyplot as plt
plt.bar(df['flight'], df['fuel_saved'], color=['red','green','orange'])
plt.title('Fuel Saved - Director Report')
plt.ylabel('KG Saved')
plt.show()

plt.bar(df['flight'], df['extra_seats'])
plt.title('Extra Seats to sell')
plt.show()
```

```python
df.to_excel('FZ_Director_Report.xlsx', index=False)
print("Director's Report Made!")
```
