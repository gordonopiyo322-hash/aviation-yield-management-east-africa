```python
from scipy.stats import norm

fare_Y=500
fare_Q=200

avg_business_demand=30
demand_bounce=5

prob_target=1-(fare_Q/fare_Y)

protected_seats=norm.ppf(prob_target, loc=avg_business_demand, scale=demand_bounce)

print("---EMSRb Revenue Management Optimization---")
print(f"Target Confidence Threshhold: {prob_target*100}%")
print(f"Recommended Seats to Protect for Business: {round(protected_seats)}")
```
