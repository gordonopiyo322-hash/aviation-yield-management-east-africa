```python
class Flight:
  def __init__(self,route="DXB-JRO", base_price=780,seats=0.80,distance_km=4500):
    self.route=route
    self.base_price=base_price
    self.seats=seats
    self.distance_km=distance_km
    self.pax_count=int(seats*200)  

  def dynamic_price(self):
      if self.seats>0.80:
        return 1300
      elif self.seats>0.50:
          return 780
      return 550

  def fuel_saving(self):
      weight_penalty=self.pax_count*100*0.05
      base_fuel_kg=self.distance_km*8.5 + weight_penalty

      optimized_fuel=base_fuel_kg*0.97
      saved_fuel_kg=base_fuel_kg - optimized_fuel

      price_per_kg=0.80
      return int(saved_fuel_kg* price_per_kg)


  def total_profit_per_seat(self):
      return self.dynamic_price() + self.fuel_saving()
      


for load in [0.95,0.85,0.40]:
     my_flight= Flight(seats=load)
     msg=f"Load {load*100:.0f}% | Price ${my_flight.dynamic_price()} | Fuel savin ${my_flight.fuel_saving()} | Total ${my_flight.total_profit_per_seat()}"
     print(msg)      
```
