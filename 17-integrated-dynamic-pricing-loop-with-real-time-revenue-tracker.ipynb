```python
TOTAL_CAPACITY = 100
current_bookings = 0
total_revenue = 0

bookings_by_class= {"Y": 0, "M":0, "Q":0}

passenger_queue= [
    ("Q", 120),
    ("Q", 110),
    ("M", 60),
    ("Q", 45),
    ("Y", 15),
    ("Y", 3),
]

print("---Live CRS Booking Engine & Revenue Counter---")

for request_class, dtd in passenger_queue:
    if current_bookings>= TOTAL_CAPACITY:
        print(f"❌Rejected : Flight is 100% full.")
        continue
        
    load_factor = current_bookings/TOTAL_CAPACITY

    if load_factor>= 0.70:
        ticket_price= 500
    else:
        if dtd > 90:
            ticket_price= 200
        elif 21 <= dtd <= 90:
            ticket_price=350
        else:
            ticket_price=500

bookings_by_class[request_class]+=1
current_bookings +=1
total_revenue +=ticket_price

print(f"✅Booked {request_class} class at DTD {dtd:3d} days. Fare: {ticket_price}")

print("\n---End of Day Financial Summary ---")
print(f"Total occupancy: {(current_bookings/TOTAL_CAPACITY)*100:.0f} seats occupied.")
print(f"Total cash collected: ${total_revenue}")


    
```
