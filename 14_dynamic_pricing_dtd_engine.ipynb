```python
def calculate_dynamic_pricing(days_to_depature, current_load_factor):
    """
    Automated Airline Pricing Engine
    Tracks DTD Calender horizons and occupancy overrides.
    """
    if current_load_factor>=0.70:
        return 500

    if days_to_depature > 90:
        return 200
    elif 21 <= days_to_depature <=90:
        return 350
    else:
        return 500

print("---Dynamic Pricing Calender Simulations---")

price_1 = calculate_dynamic_pricing(days_to_depature=120, current_load_factor=0.15)
print(f"Scenario 1 (DTD 120 , Load factor 15%): Ticket Price = ${price_1}")

price_2=calculate_dynamic_pricing(days_to_depature=45, current_load_factor=0.40)
print(f"Scenario 2 (DTD 45, Load factor 40%): Ticket Price ${price_2}")

price_3=calculate_dynamic_pricing(days_to_depature=5, current_load_factor=0.60)
print(f"Scenario 3 (DTD 5, Load factor 60%): Ticket Price ${price_3}")
      
price_4 = calculate_dynamic_pricing(days_to_depature=100, current_load_factor=0.75)
print(f"Scenario 4 (DTD 100 , Load factor 75%): Ticket Price ${price_4})<-- Override triggered!")
```
