```python
import pandas as pd
import numpy as np

demand_history=[95,102,110,98,125,130,115,120,108,132,128,135,122,130]
df=pd.DataFrame({'day':range(1,15),'passengers':demand_history})
df['forecast_3day']=df['passengers'].rolling(3).mean()
df['price_usd']=np.where(df['passengers']>120,350,250)

print(f"KESHO: {df['forecast_3day'].iloc[-1]:.0f} pax")
```
