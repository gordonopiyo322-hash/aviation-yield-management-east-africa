```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

demand=[95, 102, 110, 98, 125, 130, 115, 120, 108, 132, 128, 135, 122, 130]
df=pd.DataFrame({
    'day': range(1,15),
    'passengers': demand
})

df['forecast_3day'] = df['passengers'].rolling(window=3).mean()
df['forecast_7day'] = df['passengers'].rolling(window=7).mean()

tomorrow_forecast = df['forecast_3day'].iloc[-1]
price_tomorrow = 350 if tomorrow_forecast>120 else 250

print("---FZ1289 FORECAST TABLE---")
print(df.tail())
print(f"\nKESHO FORECAST: {tomorrow_forecast:.0f} passengers")
print(f"BEI KESHO: ${price_tomorrow}")
print(f"\nRevenue Kesho: (forecast): ${tomorrow_forecast* price_tomorrow}")


plt.figure(figsize=(12,6))
plt.plot(df['day'] , df['passengers'] ,'o-', label='Actual - Abiria halisi', linewidth=2)
plt.plot(df['day'], df['forecast_3day'] ,'s--', label= '-day Forecast - Radar haraka')
plt.plot(df['day'], df['forecast_7day'] , '^:',  label='7Day Forecast - Radar stable')
plt.axhline(120, color='red', alpha=0.3 , linestyle='-', label='Threshold ya bei')
plt.title('Fz1289 DXB-JRO Demand Forecast - Director Gordon',  fontweight='bold')
plt.xlabel('Siku')
plt.ylabel('Abiria')
plt.legend()
plt.grid(True , alpha=0.3)
plt.show()
```
