```python
class Flight:
  def __init__(self,seats,distance=4500):
    self.seats=seats
    self.distance=distance
    self.pax_count=int(seats*200)

  def fuel_saving_usd(self):
      price_per_kg=0.80
      empty_fuel=self.distance*7.5
      pax_extra_fuel=self.pax_count*100*0.05*(self.distance/100)
      actual_fuel=empty_fuel + pax_extra_fuel
      max_fuel=(self.distance*7.5) + (200*100*0.05*(self.distance/100))
      saved_by_empty_seats=max_fuel-actual_fuel
      lido_optimization=actual_fuel*0.03

      total_saved_kg=saved_by_empty_seats + lido_optimization
      return int(total_saved_kg*price_per_kg)


for load in [0.95,0.85,0.70,0.40]:
    f=Flight(load)
    print(f"Load {int(load*100)}% ({f.pax_count} pax) -> Fuel Saving = ${f.fuel_saving_usd()}")
    

```
