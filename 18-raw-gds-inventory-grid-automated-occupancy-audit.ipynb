```python
import pandas as pd

flight_matrix ={
    "Flight_No":["FZ1289", "FZ1543", "FZ1701"],
    "Route":["DXB-NBO", "DXB-AMM", "DXB-BOM"],
    "Physical_Seats":[100, 150, 180],
    "Tickets_Sold":[92, 60, 176],
    "Days_To_Depature":[12,45,2]
}

df=pd.DataFrame(flight_matrix)

print("---Step A: Raw GDS Flight Inventory Grid Loaded---")
print(df)

df["Load_Factor"] =(df["Tickets_Sold"]/df["Physical_Seats"]).round(2)

print("\n---Step B: Automated Occupancy Audit Opened---")
print(df[["Flight_No", "Route", "Days_To_Depature", "Load_Factor"]])
```
