```python
TOTAL_CAPACITY=100
PROTECTED_Y_AND_M=80
PROTECTED_Y_ONLY=31

booked_seats=0
bookings_by_class={"Y":0, "M":0, "Q":0}

passenger_requests=["Q", "Q", "Q", "M", "Q", "Y", "Q", "M", "Q", "Q", "Q"]

print("---Live Ticket Booking Capacity Check---")

for request in passenger_requests:
    if booked_seats>=TOTAL_CAPACITY:
        print(f"❌request for {request}: REJECTED-Plane is completely full!")
        continue

    if request=="Q":
        available_promo_seats = TOTAL_CAPACITY - PROTECTED_Y_AND_M
        if bookings_by_class["Q"]< available_promo_seats:
            bookings_by_class["Q"] +=1
            booked_seats +=1
            print(f"✅Request for Promo(Q): CONFIRMED (Promo sold: {bookings_by_class['Q']}/{available_promo_seats})")
        else:
            print(f"Request for Promo (Q): CLOSED - Remaining Seats protected for higher classes!")

    elif request =="M":
        available_standard_seats=TOTAL_CAPACITY - PROTECTED_Y_ONLY
        if (bookings_by_class["M"] + bookings_by_class["Q"])< available_standard_seats:
            bookings_by_class["M"] +=1
            booked_seats +=1
            print(f"✅Request for Standard(M): CONFIRMED!")
        else:
            print(f"Request for Standard (M): CLOSED - Remaining seats promoted for business class (Y)!")

    elif request =="Y":
        bookings_by_class["Y"] +=1
        booked_seats += 1
        print(f"✅Request for Business (Y): CONFIRMED")

print("\n---Final Booking Summary---")
print(f"Total Passenger Load: {booked_seats}/{TOTAL_CAPACITY} seats filled.")
print(f"Breakdown by Class: {bookings_by_class}")
            
            

```
