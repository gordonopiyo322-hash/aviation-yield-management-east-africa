```python
class AirlineRevenueSimulator:
    def __init__(self,flight_number,total_capacity):
        self.flight_number=flight_number
        self.total_capacity=total_capacity
        self.seats_remaining=total_capacity
        self.total_revenue=0

        self.prices={
         "Y":400,
         "Q":200,
         "K":100
    }

        self.booking_limits={
         "Y": total_capacity,
         "Q":4,
         "K":2
    }

        self.tickets_sold_count=0
        self.max_tickets_allowed=total_capacity + 3

    def process_booking(self,fare_code):
        if self.tickets_sold_count>=self.max_tickets_allowed:
            print(f"Flight completely maxed out! cannot sell {fare_code}.")
            return

        limit=self.booking_limits[fare_code]
        if self.tickets_sold_count>=limit:
            print(f"🚫Protection Block: Class {fare_code} is close to save seats for higher fares.")
            return

        price=self.prices[fare_code]
        self.tickets_sold_count+=1
        if self.seats_remaining>0:
            self.seats_remaining-=1

        self.total_revenue +=price
        print(f"✅ Ticket Sold! Class: {fare_code} |Price: {price} |Total Tickets Issued: {self.tickets_sold_count}")

    def final_flight_report(self,actual_no_shows):
       print(f"\n====================================")
       print(f"📊FINAL FLIGHT REPORT: {self.flight_number}")
       print(f"======================================")
       print(f"Total Tickets Sold(with overbooking): {self.tickets_sold_count}")
       print(f"Passenger who did not Show up(No-shows): {actual_no_shows}")

       boarding_passengers=self.tickets_sold_count-actual_no_shows
       print(f"Actual Passengers Boarding Gate: {boarding_passengers}")


       if boarding_passengers>self.total_capacity:
          denie=boarding_passengers-self.total_capacity
          print(f"Overbooked! {denied} Passenger(s) denied boarding(Denied boarding compensation triggered.)")

       else:
          print(f"🤑Total Revenue Generated for Airline: ${self.total_revenue}")
          print(f"=============================================")

ek_flight=AirlineRevenueSimulator("EK-505", total_capacity=5)
print("---STARTING TICKET SALES---")
ek_flight.process_booking("K")
ek_flight.process_booking("K")
ek_flight.process_booking("K")
ek_flight.process_booking("Q")
ek_flight.process_booking("Q")
ek_flight.process_booking("Y")
ek_flight.process_booking("Y")

ek_flight.final_flight_report(actual_no_shows=1)
         



```
